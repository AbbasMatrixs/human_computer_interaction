```python
import numpy as np
import matplotlib.pyplot as plt
# Original Shape (Triangle)
shape = np.array([
    [0, 0],
    [2, 0],
    [1, 2],
    [0, 0]
])

# Translation
def translate(points, tx, ty):
    T = np.array([[1, 0],
                  [0, 1]])
    return points @ T + np.array([tx, ty])

# Rotation
def rotate(points, angle):
    rad = np.radians(angle)
    R = np.array([[np.cos(rad), -np.sin(rad)],
                  [np.sin(rad),  np.cos(rad)]])
    return points @ R

# Shearing
def shear(points, shx, shy):
    Sh = np.array([[1, shx],
                   [shy, 1]])
    return points @ Sh

# Reflection
def reflect_x(points):
    Rx = np.array([[1, 0],
                   [0, -1]])
    return points @ Rx

# Apply Transformations

translated = translate(shape, 3, 2)
rotated = rotate(shape, 45)
sheared = shear(shape, 1, 0)
reflected_x = reflect_x(shape)

# Visualization

def plot_shape(ax, points, title, color):
    ax.plot(points[:,0], points[:,1], marker='o', color=color)
    ax.set_title(title)
    ax.grid(True)
    ax.axis('equal')

# ---- Individual Graphs ----
fig, axs = plt.subplots(2, 3, figsize=(12, 8))

plot_shape(axs[0,0], shape, "Original Shape", 'blue')
plot_shape(axs[0,1], translated, "Translated", 'green')
plot_shape(axs[0,2], rotated, "Rotated", 'red')
plot_shape(axs[1,0], sheared, "Sheared", 'purple')
plot_shape(axs[1,1], reflected_x, "Reflected X", 'orange')

plt.tight_layout()
plt.show()

# ---- Combined Graph ----
plt.figure(figsize=(8,6))
plt.plot(shape[:,0], shape[:,1], 'bo-', label='Original')
plt.plot(translated[:,0], translated[:,1], 'go-', label='Translated')
plt.plot(rotated[:,0], rotated[:,1], 'ro-', label='Rotated')
plt.plot(sheared[:,0], sheared[:,1], 'mo-', label='Sheared')
plt.plot(reflected_x[:,0], reflected_x[:,1], 'co-', label='Reflected X')

plt.legend()
plt.title("All Transformations Comparison")
plt.grid(True)
plt.axis('equal')
plt.show()
```
